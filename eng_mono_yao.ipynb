## Setting

```python
from pathlib import Path
import subprocess
import time
import re
import pandas as pd

# 确认你是在 repo 根目录运行 notebook
print("Current directory:", Path.cwd())

assert Path("train.py").exists(), "train.py not found. Please run this notebook from the repo root."
assert Path("shared_tokenizer/tokenizer.json").exists(), "shared_tokenizer not found."
assert Path("tiny_config.json").exists(), "tiny_config.json not found."
assert Path("small_config.json").exists(), "small_config.json not found."
assert Path("large_config.json").exists(), "large_config.json not found."

Path("outputs").mkdir(exist_ok=True)
Path("logs").mkdir(exist_ok=True)
Path("results").mkdir(exist_ok=True)

print("Setup OK.")
```

## Params

```python
CONFIG_NAME = "tiny"  # 第一次 tiny；之后改成 small / large
CONFIG_PATH = f"./{CONFIG_NAME}_config.json"

DATASET = "BabyLM-community/babylm-eng"
LANG = "eng"
RATIO = "mono"

SCALING_BUDGETS = [
    ("1M", 1_000_000),
    ("3M", 3_000_000),
    ("10M", 10_000_000),
    ("30M", 30_000_000),
    ("50M", 50_000_000),
]

MAX_LENGTH = 128
BATCH_SIZE = 32
EPOCHS = 3
LEARNING_RATE = "1e-4"
SEED = 42

print(CONFIG_NAME, CONFIG_PATH)
```

## 3 Function

```python
def run_one_model(config_name, budget_label, max_tokens):
    run_name = f"mono_eng_{budget_label.lower()}_{config_name}"
    output_dir = Path("outputs") / run_name
    log_path = Path("logs") / f"{run_name}.log"

    cmd = [
        "python", "train.py",
        "--dataset", DATASET,
        "--config", f"./{config_name}_config.json",
        "--tokenizer_dir", "./shared_tokenizer",
        "--output_dir", str(output_dir),
        "--model_name", run_name,
        "--max_length", str(MAX_LENGTH),
        "--batch_size", str(BATCH_SIZE),
        "--epochs", str(EPOCHS),
        "--learning_rate", LEARNING_RATE,
        "--max_tokens", str(max_tokens),
        "--seed", str(SEED),
    ]

    print("=" * 80)
    print("Running:", run_name)
    print("Command:")
    print(" ".join(cmd))
    print("=" * 80)

    start = time.time()

    with open(log_path, "w") as log_file:
        process = subprocess.Popen(
            cmd,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1,
        )

        output_lines = []
        for line in process.stdout:
            print(line, end="")
            log_file.write(line)
            output_lines.append(line)

        return_code = process.wait()

    end = time.time()
    runtime = round(end - start, 2)

    full_output = "".join(output_lines)

    ppl_match = re.search(r"babylm-eng: perplexity = ([0-9.]+)", full_output)
    perplexity = float(ppl_match.group(1)) if ppl_match else None

    status = "done" if return_code == 0 else "failed"

    result = {
        "Config": config_name,
        "Lang(s)": LANG,
        "Ratio": RATIO,
        "Tokens": budget_label,
        "Status": status,
        "Person": PERSON,
        "Time in s": runtime,
        "Cost": "",
        "Perplexity": perplexity,
        "Output dir": str(output_dir),
        "Log": str(log_path),
    }

    return result
```

## tiny-5 token sizes

```python
PERSON = "Yao"  # 你可以改成自己的名字

results = []

for budget_label, max_tokens in SCALING_BUDGETS:
    result = run_one_model(CONFIG_NAME, budget_label, max_tokens)
    results.append(result)

    # 每跑完一个模型就保存一次，防止中途断掉结果丢失
    df = pd.DataFrame(results)
    df.insert(0, "No.", range(1, len(df) + 1))
    df.to_csv(f"results/eng_mono_{CONFIG_NAME}_results.csv", index=False)

df
```
